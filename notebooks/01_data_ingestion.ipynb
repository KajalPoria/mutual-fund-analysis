```python
from pathlib import Path
import pandas as pd

# Project root
project_root = Path.cwd().parent
folder = project_root / "data" / "raw"

# Read all CSV files and display their information
for file in folder.glob("*.csv"):

    df = pd.read_csv(file)

    print("\n" + "=" * 80)
    print(f"File: {file.name}")
    print("=" * 80)

    print("\nShape:")
    print(df.shape)

    print("\nData Types:")
    print(df.dtypes)

    print("\nFirst 5 Rows:")
    display(df.head())

    print("\nMissing Values:")
    print(df.isnull().sum())

    print("\nDuplicate Rows:")
    print(df.duplicated().sum())

    print("\n")
```
